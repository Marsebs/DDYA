## 7) Análisis de complejidad **[5 pts]**
Incluye:
- Función T[n]
T(n)= t(n-50) +t(n-100) +t(n-200) + t(n-500) +t(n-1000) +O(1)
 
- Mejor y peor caso con gráfica

![alt text](<Multimedia (2)-2.jfif>)

```python
def vueltas_monedas(monedas, vueltas):
    if vueltas < 0:
        return float("inf")
    if vueltas == 0:
        return 0

    cantidad_minima = float("inf")
    for moneda in monedas:
        if moneda <= vueltas:
            cantidad_minima = min(
                cantidad_minima,
                1 + vueltas_monedas(monedas, vueltas - moneda),
            )
    return cantidad_minima


def memoized_vueltas(monedas, vueltas):
    if vueltas < 0:
        return float("inf")

    resultados = [float("inf")] * (vueltas + 1)
    resultados[0] = 0

    for cantidad in range(1, vueltas + 1):
        for moneda in monedas:
            if moneda <= cantidad:
                resultados[cantidad] = min(
                    resultados[cantidad],
                    1 + resultados[cantidad - moneda],
                )

    return resultados[vueltas]


monedas = [50, 100, 200, 500, 1000]

try:
    cambio = int(input("Ingrese el valor numérico del cambio: "))

    if cambio <= 0:
        print("Por favor ingrese un valor positivo.")
    else:
        print("El cambio es:", cambio)
        cantidad = memoized_vueltas(monedas, cambio)

        if cantidad == float("inf"):
            print("No es posible entregar el cambio exacto.")
            print("Por favor ingrese otro valor.")
        else:
            print("Cantidad mínima de monedas:", cantidad)

except ValueError:
    print("Por favor ingrese un valor numérico entero.")
```

## 9) pruebas **[15 pts]**

```python
def test(entrada, salida_esperada):
    salida_obtenida = memoized_vueltas(monedas, int(entrada))
    salida_esperada = int(salida_esperada)

    assert salida_obtenida == salida_esperada, (
        f"Se obtuvo {salida_obtenida}, pero se esperaba {salida_esperada}"
    )

    print(f"Pasó caso de prueba {entrada} -> {salida_obtenida}")
```

#### Código base para benchmarking

```python
# Calcula el tiempo de ejecución de memoized_vueltas
# con diferentes tamaños de entrada.

import time


def bench(n, step=50, repeticiones=20):
    if n < 50:
        raise ValueError("n debe ser mayor o igual a 50.")
    if step <= 0:
        raise ValueError("step debe ser un número positivo.")
    if repeticiones <= 0:
        raise ValueError("repeticiones debe ser un número positivo.")

    print("Generando datos ...", end="")

    resultados = []
    monedas_bench = [50, 100, 200, 500, 1000]

    # Las cantidades de cambio serán múltiplos de 50.
    for size in range(50, n + 1, step):
        inicio = time.perf_counter()

        # Se repite la medición para reducir el ruido de tiempos muy pequeños.
        for _ in range(repeticiones):
            memoized_vueltas(monedas_bench, size)

        fin = time.perf_counter()
        tiempo_promedio_ms = ((fin - inicio) / repeticiones) * 1000

        print(".", end="")
        resultados.append((size, tiempo_promedio_ms))

    print()
    return resultados


# === CÓDIGO PARA DIBUJAR ===

def dibujar(x, y, dimensiones=(7, 5)):
    if len(x) != len(y):
        raise ValueError("x e y deben tener la misma cantidad de datos.")

    import matplotlib.pyplot as plt

    plt.figure(figsize=dimensiones)
    plt.plot(x, y, marker="o")
    plt.xlabel("Valor del cambio")
    plt.ylabel("Tiempo promedio de ejecución (ms)")
    plt.title("Tiempo de ejecución de memoized_vueltas")
    plt.grid(True)
    plt.tight_layout()
    plt.show()


# === LLAMADO ===

tiempos = bench(5000, step=50, repeticiones=20)

x = [size for size, _ in tiempos]
y = [tiempo for _, tiempo in tiempos]

dibujar(x, y)
```
